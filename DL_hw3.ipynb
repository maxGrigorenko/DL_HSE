```python
import os
import torch
from PIL import Image
from IPython.display import display, Image as IPImage, Markdown
from transformers import AutoModelForImageTextToText, AutoProcessor
from qwen_utils import process_vision_info

import warnings
warnings.filterwarnings('ignore')

from tam import TAM

# 1. Настройка параметров
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
MODEL_NAME = "OpenGVLab/InternVL3-2B-hf"
IMAGE_PATH = "my_tests/test_image.jpg"
PROMPT = "Who is depicted in the picture?"
SAVE_DIR = "my_tests/results"

os.makedirs(SAVE_DIR, exist_ok=True)
```

```python
torch_device = "cuda" if torch.cuda.is_available() else "cpu"
processor = AutoProcessor.from_pretrained(MODEL_NAME)

model = AutoModelForImageTextToText.from_pretrained(
    MODEL_NAME,
    device_map=torch_device,
    torch_dtype=torch.bfloat16
)
```

```python
# 3. Подготовка входных данных (Специфика InternVL)
# InternVL в демо авторов требует жесткого ресайза до 448x448
image = Image.open(IMAGE_PATH).convert('RGB')
image = image.resize((448, 448))

conversation = [{"role": "user", "content": [{"type": "image"}, {"type": "text", "text": PROMPT}]}]
text_prompt = processor.apply_chat_template(conversation, add_generation_prompt=True)

inputs = processor(text=[text_prompt], images=[image], padding=True, return_tensors="pt")
inputs = inputs.to(model.device).to(model.dtype)

# 4. Генерация
print("Генерация ответа и анализ внимания...")
outputs = model.generate(
    **inputs,
    max_new_tokens=50,
    output_hidden_states=True,
    return_dict_in_generate=True
)

generated_ids = outputs.sequences
logits = [model.lm_head(feats[-1]) for feats in outputs.hidden_states]

gen_tokens = generated_ids[0][-len(logits):]

# 5. Специфика TAM для InternVL (взято из исходников авторов)
vision_shape = (16, 16)
vis_inputs = image
special_ids = {
    'img_id': [151665, 151666],
    'prompt_id': [[151666, 198], [151645, 198, 151644, 77091]],
    'answer_id': [[198, 151644, 77091, 198], -1]
}

# 6. Визуализация в стиле Colab
raw_map_records = []
print("\n" + "="*40)
print("ПОСЛОВНЫЙ РАЗБОР ВНИМАНИЯ (InternVL3-1B)") # Забыл исправить, тут должно быть 2B
print("="*40 + "\n")

for i in range(len(logits)):
    save_path = os.path.join(SAVE_DIR, f"{i}.jpg")

    TAM(
        generated_ids[0].cpu().tolist(),
        vision_shape,
        logits,
        special_ids,
        vis_inputs,
        processor,
        save_path,
        i,
        raw_map_records,
        False
    )

    token_str = processor.decode([gen_tokens[i]])

    if os.path.exists(save_path):
        display(Markdown(f"### Сгенерировано: **`{token_str}`** (Токен #{i})"))
        display(IPImage(filename=save_path))
        print("-" * 40)
```

### Вывод: Анализ ошибки модели (Failure Case Analysis) с помощью алгоритма TAM

В ходе тестирования алгоритма TAM на легковесной модели семейства **InternVL** был проведен анализ генерации ответа по фотографии Фридриха Ницше. Модель допустила фактическую ошибку, идентифицировав человека на портрете как «Фридрих Энгельс» (Friedrich Engels).

Построение пословных тепловых карт внимания (TAM) позволило вскрыть механику этой ошибки и доказать, что она вызвана **доминированием языковой модели** над визуальным энкодером:

1. **Успешное первичное распознавание (Токен `Friedrich`):** При генерации первого токена алгоритм TAM зафиксировал высокую визуальную активацию. Фокус внимания модели был четко направлен на лицо и характерные усы исторического деятеля. Модель правильно считала первичный визуальный образ.
2. **Затухание внимания и галлюцинация (Токены `Eng` и `els`):** При генерации последующих токенов фамилии визуальное внимание на тепловых картах практически исчезает. Модель перестает опираться на изображение.
3. **Причина сбоя:** Сгенерировав слово «Фридрих» в контексте фотографии мужчины 19 века, авторегрессионная языковая часть модели (LLM) перехватила управление. Скорее всего, из-за сильной статистической связи в обучающих корпусах текстов, модель по инерции дописала более частотную фамилию «Энгельс», полностью проигнорировав тот факт, что визуально Ницше и Энгельс не похожи.

**Итог:** Данный эксперимент наглядно подтверждает ценность алгоритма TAM для разработчиков ИИ. Визуализация доказала, что сбой произошел не из-за "слепоты" визуального энкодера, а из-за того, что языковая модель проигнорировала визуальные признаки в пользу своего текстового контекста (возникла галлюцинация). Для сравнения, архитектура Qwen2-VL на 2B параметров в аналогичном тесте показала лучший баланс между зрением и текстом, корректно распознав Ницше (но в финальной версии я этот блок убрал, чтобы протестировать другую модель).
